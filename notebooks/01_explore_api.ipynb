```python
import json
import requests

r = requests.get(
    "https://transport.opendata.ch/v1/stationboard",
    params={"id": "8504300", "limit": 2},   # Biel/Bienne, 2 departures
    timeout=15,
)
r.raise_for_status()
first = r.json()["stationboard"][0]
print(json.dumps(first, indent=2, ensure_ascii=False))
```

```python
r = requests.get(
    "https://transport.opendata.ch/v1/stationboard",
    params={"id": "8504300", "limit": 10},
    timeout=15,
)
for dep in r.json()["stationboard"]:
    print(
        dep["category"], dep["number"], "->", dep["to"],
        "| arrival at Biel:", dep["stop"]["arrival"],
        "| first stop:", dep["passList"][0]["station"]["name"],
    )
```

```python
dep = r.json()["stationboard"][9]          # the IC 5 -> Lausanne
for s in dep["passList"][:4]:
    print(s["station"]["id"], s["station"]["name"], s["departure"])
```

```python
data  = r.json()                    # dict, the whole answer
board = data["stationboard"]        # list of trains
dep   = board[2]                    # dict, one train (IC 5 -> Lausanne)
stop  = dep["stop"]                 # dict, its stop at Biel
print(stop["departure"])            # string: 2026-10-04T21:18:00+0200
print(dep["passList"][3]["station"]["name"])   # Neuchâtel
```

```python
from datetime import datetime, timezone

dep = r.json()["stationboard"][2]
departure_utc = datetime.fromtimestamp(dep["stop"]["departureTimestamp"], tz=timezone.utc)
row = {
    "station_id":            dep["stop"]["station"]["id"],
    "category":              dep["category"],
    "line_number":           dep["number"],
    "operator":              dep["operator"],
    "destination":           dep["to"],               # done
    "departure_planned_utc": departure_utc,
    "departure_planned_raw": dep["stop"]["departure"],  # done
    "delay_min":             dep["stop"]["delay"],    # done
    "platform":              dep["stop"]["platform"],
}
print(row)
```
